# OCTA500_6M

```python
import autorootcwd
import numpy as np
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt

data_list = {
    "train/image": "data/OCTA500_6M/train/images/10001.bmp",
    "train/CSNet_input": "data/OCTA500_6M/CSNet/train/batch_0_image.jpg",
    "train/FRUnet_input": "data/OCTA500_6M/FRUnet/train/batch_0_image.jpg",
    "train/SwinUNETR_input": "data/OCTA500_6M/SwinUNETR/train/batch_0_image.jpg",
    "train/Unet_input": "data/OCTA500_6M/Unet/train/batch_0_image.jpg",
    "train/label": "data/OCTA500_6M/train/labels/10001.bmp",
    "train/CSNet_target": "data/OCTA500_6M/CSNet/train/batch_0_mask.jpg",
    "train/FRUnet_target": "data/OCTA500_6M/FRUnet/train/batch_0_mask.jpg",
    "train/SwinUNETR_target": "data/OCTA500_6M/SwinUNETR/train/batch_0_mask.jpg",
    "train/Unet_target": "data/OCTA500_6M/Unet/train/batch_0_mask.jpg",
    "test/image": "data/OCTA500_6M/test/images/10201.bmp",
    "test/CSNet_input": "data/OCTA500_6M/CSNet/test/batch_0_image.jpg",
    "test/FRUnet_input": "data/OCTA500_6M/FRUnet/test/batch_0_image.jpg",
    "test/SwinUNETR_input": "data/OCTA500_6M/SwinUNETR/test/batch_0_image.jpg",
    "test/Unet_input": "data/OCTA500_6M/Unet/test/batch_0_image.jpg",
    "test/label": "data/OCTA500_6M/test/labels/10201.bmp",
    "test/CSNet_target": "data/OCTA500_6M/CSNet/test/batch_0_mask.jpg",
    "test/FRUnet_target": "data/OCTA500_6M/FRUnet/test/batch_0_mask.jpg",
    "test/SwinUNETR_target": "data/OCTA500_6M/SwinUNETR/test/batch_0_mask.jpg",
    "test/Unet_target": "data/OCTA500_6M/Unet/test/batch_0_mask.jpg",
}
plt.figure(figsize=(10, 6))
for index, key in enumerate(data_list.keys()):
    file = data_list[key]
    data = open(file, "rb").read()
    data = Image.open(BytesIO(data))
    data_npy = np.array(data)
    print(f"{key}: {data_npy.shape}, {data_npy.dtype}, mean: {data_npy.mean():.2f} ")
    print(f"min: {data_npy.min()}, max: {data_npy.max()}, # of unique values : {len(np.unique(data_npy))}")
    plt.subplot(4, 5, list(data_list.keys()).index(key) + 1)
    plt.title(key, fontsize=5)
    plt.imshow(data_npy, cmap='gray')
    plt.axis('off')
```

## Results

```python
import autorootcwd
from matplotlib import pyplot as plt
import numpy as np
from PIL import Image
from io import BytesIO

data_list = {
    "image": "workdir/poisson-OCTA500_6M-FRUnet/results-050000/0_image_0.png",
    "input": "workdir/poisson-OCTA500_6M-FRUnet/results-050000/0_input_0.png",
    "pred": "workdir/poisson-OCTA500_6M-FRUnet/results-050000/0_final_output_0.png",
    "output": "workdir/poisson-OCTA500_6M-FRUnet/results-050000/0_output_0.png",
    "target": "workdir/poisson-OCTA500_6M-FRUnet/results-050000/0_target_0.png"
}
plt.figure(figsize=(12, 8))
for key in data_list.keys():
    file = data_list[key]
    data = open(file, "rb").read()
    data = Image.open(BytesIO(data))
    data_npy = np.array(data)
    print(f"{key}: {data_npy.shape}, {data_npy.dtype}, mean: {data_npy.mean():.2f} ")
    print(f"min: {data_npy.min()}, max: {data_npy.max()}, # of unique values : {len(np.unique(data_npy))}")
    float_values = data_npy.astype(np.float32) / 255.0
    print(f"After scaling to [0, 1]: min: {float_values.min()}, max: {float_values.max()}, # of unique values : {len(np.unique(float_values))}")
    print()
    plt.subplot(2, 3, list(data_list.keys()).index(key) + 1)
    plt.imshow(float_values, cmap='gray')
    plt.title(f"{key}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```
